```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load your CSV (or use pd.read_clipboard if you're pasting it directly)
df = pd.read_csv("leave_one_out_city_china_test_on_season_SEASON 1.csv")
df = df[df['n_tasks'] == 4]

# Plot: Grouped bar plot by leave_out_task (season), each bar is a method
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x="leave_out_task", y="loss", hue="method")

plt.title("Loss by Method for Each Leave-Out Task - Season 1")
plt.xlabel("Leave-Out Task (Season)")
plt.ylabel("Loss")
plt.legend(title="Method", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load your data
df = pd.read_csv("leave_one_out_city_china_test_on_season_SEASON 1.csv")

# Create pivot tables
methods = df['method'].unique()
n_methods = len(methods)

# Set up figure and axes
fig, axes = plt.subplots(1, n_methods, figsize=(5 * n_methods, 6), sharey=True)

# Flatten axes if only one method
if n_methods == 1:
    axes = [axes]

# Compute global min/max loss for consistent color scale
vmin = df['loss'].min()
vmax = df['loss'].max()

# Create the heatmaps
for i, method in enumerate(methods):
    data_method = df[df['method'] == method]
    pivot = data_method.pivot(index="leave_out_task", columns="n_tasks", values="loss")
    
    # Only show colorbar on the last subplot
    sns.heatmap(pivot, ax=axes[i], annot=True, fmt=".3f", cmap="mako", 
                vmin=vmin, vmax=vmax, cbar=(i == n_methods - 1),
                cbar_kws={"label": "Loss"} if i == n_methods - 1 else None)
    
    axes[i].set_title(method)
    axes[i].set_xlabel("n_tasks")
    if i == 0:
        axes[i].set_ylabel("Leave-Out Task")
    else:
        axes[i].set_ylabel("")

plt.tight_layout()
plt.show()
```
